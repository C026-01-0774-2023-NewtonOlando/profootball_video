```python
from ultralytics import YOLO
import yaml
import os

# Paths
model_path = "/content/part2.pt"
data_path = "/content/Football-Players-ball-referee-detection-1/data.yaml"

# Check files
print("Model exists:", os.path.exists(model_path))
print("Dataset exists:", os.path.exists(data_path))

# Load model
model = YOLO(model_path)

print("\nModel classes:")
print(model.names)

# Read dataset YAML
with open(data_path, "r") as f:
    data = yaml.safe_load(f)

print("\nDataset classes:")
print(data["names"])

print("\nNumber of classes:", data["nc"])
```

```python
import torch

print("PyTorch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
print("GPU count:", torch.cuda.device_count())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
results = model.train(
    data="/content/Football-Players-ball-referee-detection-1/data.yaml",

    epochs=15,
    imgsz=1280,
    batch=4,

    # Small learning rate so we don't destroy the improvements
    # from the previous training
    lr0=0.00005,
    lrf=0.01,

    optimizer="AdamW",
    weight_decay=0.0001,

    device=0,
    amp=True,

    # Moderate augmentation
    degrees=5.0,
    translate=0.1,
    scale=0.5,
    fliplr=0.5,

    # Stop if validation stops improving
    patience=5,

    # Save this experiment separately
    project="/content/football_training",
    name="football_player_recovery",
    exist_ok=True,

    save=True,
    plots=True,
    verbose=True
)
```
